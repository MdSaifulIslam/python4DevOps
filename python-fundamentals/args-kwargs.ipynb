# Flexible Functions: `*args` and `**kwargs`

- We can use the syntax `*args` and `**kwargs` to accept a variable number of both positional and keyword arguments.

```python
def example_function(*args, **kwargs):
    print(f"Positional args: {args[1]}")
    print(f"Keyword args: {kwargs['a']}") # it  can access by index

example_function(1, 2, 3, a="first", b=True)
#example_function(1, 2, 3) error, there is no kwargs and index 'a' missing
```

## `*args` in Definition: Collecting Positionals
- Uses `*args` to gather extra positional parameters into a tuple
- Allows functions to accept any number of positional inputs
- Common in utilities like custom logging or aggregation functions

```python
def apply_operator(operator, *operands):
    """
        Apply operator to a variable number of operands. Supports 'add' and 'mul'

        Args:
            operator (str): The operator to apply. Must be either 'add' or 'mul'.
            operands (int or float): Zero or more numbers to be combined.

        Returns:
            int or float: The result of applying operator on the operands.

        Raises:
            ValueError: Raised when operator is not 'add' nor 'mul'
    """

    if operator == 'add':
        result = sum(operands)
    elif operator == 'mul':
        result = 1
        for n in operands:
            result *= n
    else:
        raise ValueError(f"Unknown operator '{operator}'. Supported values are 'add' and 'mul'")
    
    return result

print(apply_operator('add', 1, 2, 3, 4, 5))
print(apply_operator('mul', 1, 2, 3, 4, 5))
print(apply_operator('mul'), apply_operator('add'))
# print(apply_operator('div', 1, 2, 3, 4, 5)) raises ValueError, 'div' is not supported
```

## `**kwargs` in Definition: Collecting Keywords
- Uses `**kwargs` to gather extra named parameters into a dictionary
- Ideal for optional configuration flags or settings
- Enables functions to accept flexible keyword arguments without predefining them

```python
def set_options(**settings):
    print(f"Received dictionary/settings: {settings}")
    for key, itam in settings.items():
        print(f"\t{key} = {itam}")

set_options(user='Admin', timeout=400, retry=5)
```

## Order in Definition Matters
- Standard positional parameters must come first, some might also have a default value
- Followed by `*args` to catch extra positionals
- Then keyword-only parameters, some might also have a default value
- Finally `**kwargs` to catch extra keyword arguments

```python
def process_request(url, method='GET', *headers, timeout, **params):
    print(f"url = {url}, method = {method}, timeout = {timeout}")
    print(f"headers = {headers}")
    print(f"params = {params}")

process_request("http://example.com", 'POST', 1, 2, 3, 4, timeout=300, user='Admin', role='All')
process_request("http://example.com", 'GET', 'Auth=JWT', 'Content_type=application/json', timeout=300, user='Admin', role='All')
process_request("http://example.com", 'POST', 1, 2, 3, Auth='JWT', Content_type='application/json', timeout=300, user='Admin', role='All') # Auth and content_type considered as params
# process_request("http://example.com", 'POST', Auth='JWT', Content_type='application/json' timeout=300, user='Admin', role='All') # Auth and content_type considered as params
# process_request("http://example.com", 'POST', timeout=300, 1, 2, 3, 4,  user='Admin', role='All') raises SyntaxError: positional argument follows keyword argument
```

## `*` in Call: Unpacking Positional Arguments
- Uses `*sequence` to expand a list or tuple into positional arguments
- Sequence length must match the function’s positional parameters
- Useful for dynamic argument lists built at runtime

```python
def connect(host, port, timeout, *args): # *args for extra safety
    print(f"Connecting to {host}:{port} with timeout {timeout}")

params = ['db.internal', 3306, 300]
connect(*params)
params = ['db.internal', 3306, 300, 'a', 'a_is_extra'] # TypeError: connect() takes 3 positional arguments but 5 were given, sotution -> *params[:3]
connect(*params[:3])
```

## `**` in Call: Unpacking Keyword Arguments
- Uses `**dict` to expand key-value pairs into keyword arguments
- Dictionary keys must match the function’s parameter names
- Common in configuration-driven function calls

```python
def configure_service(name, version, replicas=1):
    print(f"Setting up {name} v{version} with {replicas} replicas ...")

config = {"name": "auth-service", "replicas": 3, "version": "2.1.0"}

configure_service(*config.values()) # auth-service, 3, 2.1.0
configure_service(**config) # name=auth-service, replicas=3, version=2.1.0
```
